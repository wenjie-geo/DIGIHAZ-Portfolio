```python
import json
from pathlib import Path

print("Libraries imported successfully.")
```

```python
class GeoJSONManager:
    """
    A simple class for reading, writing, creating, and manipulating GeoJSON files.

    This class is designed for Assignment 3:
    - Read a GeoJSON from a file
    - Write a GeoJSON to a file
    - Create a new GeoJSON FeatureCollection
    - Add Polygon features
    """

    def __init__(self):
        self.data = None

    def create_new_geojson(self):
        """
        Create an empty GeoJSON FeatureCollection.
        """
        self.data = {
            "type": "FeatureCollection",
            "features": []
        }
        return self.data

    def read_geojson(self, file_path):
        """
        Read a GeoJSON file from disk.
        """
        file_path = Path(file_path)

        with open(file_path, "r", encoding="utf-8") as f:
            self.data = json.load(f)

        return self.data

    def write_geojson(self, file_path):
        """
        Write the current GeoJSON data to disk.
        """
        if self.data is None:
            raise ValueError("No GeoJSON data available. Please create or read GeoJSON first.")

        file_path = Path(file_path)

        with open(file_path, "w", encoding="utf-8") as f:
            json.dump(self.data, f, indent=2, ensure_ascii=False)

        print(f"GeoJSON file written successfully: {file_path}")

    def add_polygon_feature(self, name, coordinates, properties=None):
        """
        Add a Polygon feature to the GeoJSON FeatureCollection.

        Parameters
        ----------
        name : str
            Name of the polygon feature.
        coordinates : list
            Polygon coordinates in GeoJSON format.
            Example:
            [
                [
                    [longitude, latitude],
                    [longitude, latitude],
                    [longitude, latitude],
                    [longitude, latitude],
                    [longitude, latitude]
                ]
            ]
        properties : dict, optional
            Additional attribute information.
        """
        if self.data is None:
            self.create_new_geojson()

        if properties is None:
            properties = {}

        properties["name"] = name

        feature = {
            "type": "Feature",
            "geometry": {
                "type": "Polygon",
                "coordinates": coordinates
            },
            "properties": properties
        }

        self.data["features"].append(feature)
        return feature

    def list_features(self):
        """
        Print a summary of all features in the GeoJSON.
        """
        if self.data is None:
            print("No GeoJSON data loaded.")
            return

        features = self.data.get("features", [])

        print(f"Number of features: {len(features)}")

        for i, feature in enumerate(features, start=1):
            geometry_type = feature.get("geometry", {}).get("type", "Unknown")
            name = feature.get("properties", {}).get("name", "Unnamed")
            print(f"{i}. {name} - {geometry_type}")
```

```python
manager = GeoJSONManager()

manager.create_new_geojson()

sample_polygon = [
    [
        [100.0, 0.0],
        [101.0, 0.0],
        [101.0, 1.0],
        [100.0, 1.0],
        [100.0, 0.0]
    ]
]

manager.add_polygon_feature(
    name="Sample Polygon",
    coordinates=sample_polygon,
    properties={
        "description": "A simple sample polygon for testing GeoJSON writing."
    }
)

manager.write_geojson("sample.geojson")
manager.list_features()
```

```python
reader = GeoJSONManager()

reader.read_geojson("sample.geojson")
reader.list_features()
```
